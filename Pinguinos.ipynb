Desafortunadamente, no han podido registrar todas las especie de pingüino, pero saben que existen **al menos tres**.

¡Utiliza tus habilidades de aprendizaje no supervisado para agrupar en clústeres el conjunto de datos de pingüinos!

Importa, investiga y preprocesa el conjunto de datos "penguins.csv".
Realiza un análisis de clústeres basado en un número razonable de clústeres y recopila sus valores promedio. El resultado debe ser un DataFrame que muestre la media de las variables originales (o columnas en "penguins.csv") por clúster.

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

penguins_df = pd.read_csv('penguins.csv')


penguins_clean = penguins_df.dropna()

penguins_preprocessed = pd.get_dummies(penguins_clean)

scaler = StandardScaler()
X = scaler.fit_transform(penguins_preprocessed)

n_clusters = 3
kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
penguins_clean['cluster'] = kmeans.fit_predict(X)


stat_penguins = penguins_clean.groupby('cluster').mean(numeric_only=True)

display(stat_penguins)
```
