### Ejercicio 3

Utilizando las técnicas que conoces hasta ahora, construye un modelo KMeans para la base de datos "wine.csv". La base de datos contiene mediciones de algunas propiedades de vinos provenientes de un viñedo de California.

La columna class_name de la tabla contiene el tipo de vino del que se toma la muestra. El resto de las columnas corresponden a diferentes propiedades del vino.

Crea e imprime la tabla cruzada para evaluar el modelo que obtuviste.
Basado en la tabla, ¿Qué tan bueno es el modelo KMeans que construiste?

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
ks = range(1, 6)
inercias = []
dfw = pd.read_csv('wine.csv')
x = dfw.drop('class_name', axis=1)
dfw.head()
y = dfw['class_name']

for k in ks:
    # Crear una instancia de KMeans con k clústeres: model
    modelo = KMeans(n_clusters=k)

    # Ajustar el modelo a las muestras
    modelo.fit(x)

    # Añadir la inercia a la lista de inercias
    inercias.append(modelo.inertia_)

# Graficar ks vs. inercias
plt.plot(ks, inercias, '-o')
plt.xlabel('Número de clusters, k')
plt.ylabel('Inercia')
plt.xticks(ks)
plt.show()

# Crear un modelo KMeans con la cantidad de clusters apropiada
modelo = KMeans(n_clusters=3)

# Ajustar el modelo a las muestras
predicciones = modelo.fit_predict(x)

# Crear un DataFrame con etiquetas y variedades como columnas: df
dfw = pd.DataFrame({'Vino': dfw["class_name"], 'variedades': predicciones})

# Crear tabla cruzada: ct
ct = pd.crosstab(dfw['Vino'], dfw['variedades'])

# Mostrar ct
print(ct)
```
